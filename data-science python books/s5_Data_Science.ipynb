```python

import matplotlib.pyplot as plt
import numpy as np

# Create sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)


# Plotting with varying linewidths
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sin(x)', linewidth=1)
plt.plot(x, y2, label='cos(x)', linewidth=3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Varying Linewidths')
plt.legend()
plt.grid(True)
plt.show()


# Plotting with varying colors
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sin(x)', color='red')
plt.plot(x, y2, label='cos(x)', color='blue')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Varying Colors')
plt.legend()
plt.grid(True)
plt.show()


# Plotting with varying linestyles
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sin(x)', linestyle='-')
plt.plot(x, y2, label='cos(x)', linestyle='--')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Varying Linestyles')
plt.legend()
plt.grid(True)
plt.show()

# Plotting with different markers

plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sin(x)', marker='o', linestyle='none') # linestyle='none' to show only markers
plt.plot(x, y2, label='cos(x)', marker='s', linestyle='none')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Different Markers')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data for demonstration (replace with your actual datasets)
np.random.seed(42)  # for reproducibility
data1 = np.random.normal(0, 1, 100)
data2 = np.random.normal(2, 1.5, 120)
data3 = np.random.normal(-1, 0.5, 80)
data4 = np.random.normal(3, 0.8, 150)

data = [data1, data2, data3, data4]
labels = ['Dataset 1', 'Dataset 2', 'Dataset 3', 'Dataset 4']
colors = ['skyblue', 'lightcoral', 'lightgreen', 'gold']

plt.figure(figsize=(10, 6))
box = plt.boxplot(data,
                  notch=True, vert=True, patch_artist=True,
                  labels=labels)


for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)

# Change color and linewidth of whiskers
for whisker in box['whiskers']:
    whisker.set(color='#7570b3', linewidth=1.5)

# Change color and linewidth of caps
for cap in box['caps']:
    cap.set(color='#7570b3', linewidth=2)

# Change color and linewidth of medians
for median in box['medians']:
    median.set(color='black', linewidth=2)

# Change style of fliers
for flier in box['fliers']:
    flier.set(marker='D', color='#e7298a', alpha=0.5)

# Add labels (customize as needed)
plt.xlabel('Datasets')
plt.ylabel('Values')
plt.title('Box Plots of Four Datasets')


plt.show()
```

```python

import matplotlib.pyplot as plt

hr_salaries = [3500, 4000, 4500, 4200, 3800]
marketing_salaries = [5000, 5500, 5200, 4800, 5300]
it_salaries = [7000, 7500, 7100, 6900, 7300]

median_hr = sorted(hr_salaries)[len(hr_salaries) // 2]
median_marketing = sorted(marketing_salaries)[len(marketing_salaries) // 2]
median_it = sorted(it_salaries)[len(it_salaries) // 2]

data = [hr_salaries, marketing_salaries, it_salaries]
plt.figure(figsize=(8, 6))
plt.boxplot(data, labels=['HR', 'Marketing', 'IT'], patch_artist=True,
            boxprops=dict(facecolor='lightblue', color='black'),
            flierprops=dict(marker='o', color='red', markersize=6),
            showmeans=True, meanline=True, meanprops=dict(linestyle='--', linewidth=2, color='orange'))

plt.text(1, median_hr, f'{median_hr}', ha='center', va='bottom', color='black')
plt.text(2, median_marketing, f'{median_marketing}', ha='center', va='bottom', color='black')
plt.text(3, median_it, f'{median_it}', ha='center', va='bottom', color='black')

plt.title('Salary Distribution by Department')
plt.xlabel('Department')
plt.ylabel('Salary (USD)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def func(x, y):
    return np.exp(x) * np.cos(y)

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = func(X, Y)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

fig.colorbar(surf)

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title(r'3D Surface Plot of $z = e^x \cdot \cos(y)$')

plt.show()
```
